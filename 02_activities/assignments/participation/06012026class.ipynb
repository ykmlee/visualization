```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x=np.arange(50)
y=np.random.randint(0,100,50)
```

```python
fig,ax=plt.subplots(figsize=(5,3)) #height is 5 inchs; weight is 3 inches
ax.scatter(x,y)
```

```python
fig,ax=plt.subplots(figsize=(5,3)) #height is 5 inchs; weight is 3 inches
ax.bar(x,y)
```

```python
fig,ax=plt.subplots(figsize=(5,3)) #height is 5 inchs; weight is 3 inches
ax.plot(x,y)
```

```python
fig,ax=plt.subplots(figsize=(5,3)) #height is 5 inchs; weight is 3 inches
ax.hist(y) #only y, not x
```

```python
fig,ax=plt.subplots(figsize=(5,3)) #height is 5 inchs; weight is 3 inches
ax.plot(x,y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()
```

```python
font1 = {'family':'sans-serif','color':'blue','size':20}
font2 = {'family':'monospace','color':'green','size':14}
fig,ax=plt.subplots(figsize=(5,3)) #height is 5 inchs; weight is 3 inches
ax.plot(x,y)
ax.set_title('Total growth over time',fontdict=font1, loc='right')
ax.set_ylabel('Total growth',fontdict=font2)
ax.set_xlabel('Years since start',fontdict=font1)
fig.tight_layout()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,
           y,
           marker='*', # NEW ADDITIONS
           color = "indigo")  # NEW ADDITIONS

fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,
           y,
           marker='*',
           color = "indigo",
           linestyle='--',  # NEW ADDITIONS, without it plot default is continuous line
           linewidth =2)  # NEW ADDITIONS, thickens lines

fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
# color being the new addition

ax.plot(x,y,marker='*', color = '#7425b9', linestyle = '--', linewidth = 2)

fig.show()
```

```python
ax.plot(x,y,marker='*',
        markersize = 12, # NEW ADDITIONS
        color = '#7425b9',
        linestyle = '--',
        linewidth = 2,
        markeredgecolor = '#fa9359', # NEW ADDITIONS
        markerfacecolor = '#000000' ) # NEW ADDITIONS
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.grid(axis = 'y', color='blue',linewidth=2,linestyle='-.')

ax.plot(x,y,marker='*',
        markersize = 12, # NEW ADDITIONS
        color = '#7425b9',
        linestyle = '--',
        linewidth = 2,
        markeredgecolor = '#fa9359', # NEW ADDITIONS
        markerfacecolor = '#000000' ) # NEW ADDITIONS
```
