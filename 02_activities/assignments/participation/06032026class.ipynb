```python
# library & dataset
import seaborn as sns
df = sns.load_dataset('iris')

# plot
sns.violinplot(x=df["species"], y=df["sepal_length"])
```

> Asthetic (Is the visualization pleasing to look at? ) - Yes the plots simple, easy to look at
> Substantive (Does the visualization accurately and honestly present data?) - Yes, it shows the distribution. The box and whisker plots accurately present data.
> Perceptual (Can we understand what message the maker of the visualization is attempting to convey?) - Yes, the violin plots shows the spread of data effectively.
